```python
import pandas as pd
from pandas.api.types import CategoricalDtype
```

```python
def proj(pd_array):
    ar = pd.Series(pd_array)
    codec = ar.unique()
    keys = ar.astype(CategoricalDtype(codec)).cat.codes
    return (list(keys), list(codec))
```

```python
def key(pd_array):
    return proj(pd_array)[0]
```

```python
def ass(*ar):
    return key(zip(*ar))
```

```python
def inv(key):
    compteur = [0] * (max(key) + 1)
    inverse = [0] * len(key)
    for i in range(len(key)):
        inverse[i] = compteur[key[i]]
        compteur[key[i]] += 1
    return inverse   
    
```

```python
def n(key):
    return len(key)
def x(key):
    return max(key) + 1
def ma(key):
    return max(key)
def d(key):
    return max(inv(key))
def mi(key):
    return min(inv(key))
```

```python
tst = ['e', 2, 'e', 1]
key(tst)
```
